```python
import pandas as pd

# Load the Excel file
df = pd.read_csv("data/Dataset_Health.csv")

# Display first few rows to check the data
df.head()
```

```python
steps_columns = [col for col in df.columns if "Steps" in col]
df_steps = df[steps_columns]

import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
for i in range(5):
    plt.plot(df_steps.iloc[i].values, label=f'Patient {i+1}')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.title('Step Trajectories of First 5 Patients')
plt.legend()
plt.show()
```

```python
import numpy as np

# Map days to weekdays; assume col names are like Steps_Day_1 ... Steps_Day_30 in order
num_days = len(steps_columns)
weekdays = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_indices = [i % 7 for i in range(num_days)]

plt.figure(figsize=(12, 7))

# For each day of the week (0 = Monday, ..., 6 = Sunday)
for wd_ix, weekday in enumerate(weekdays):
    day_cols = [steps_columns[i] for i in range(num_days) if weekday_indices[i] == wd_ix]
    plt.clf()
    plt.figure(figsize=(8,5))
    for patient_ix in range(5):
        steps_on_selected_weekday = df.loc[patient_ix, day_cols].values
        plt.plot(range(1, len(steps_on_selected_weekday)+1), steps_on_selected_weekday, label=f'Patient {patient_ix+1}')
    plt.xlabel(f"{weekday} Occurrence")
    plt.ylabel("Steps")
    plt.title(f"Steps on {weekday} (First 5 Patients)")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
